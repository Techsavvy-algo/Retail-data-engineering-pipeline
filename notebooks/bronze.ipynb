```python
from pyspark.sql.functions import current_timestamp
```

```python
raw_df = (
    spark.read.format('csv')\
        .option('header', True)\
            .option('inferSchema', True)\
                .load("/Volumes/workspace/default/my_volume")
)
```

```python
raw_df.printSchema()
```

```python
raw_df.display()
```

```python
raw_df.count()
```

#### Creting the BRONZE DATAFRAME - So adding the ingestion Timestamp

```python
bronze_df = raw_df.withColumn('ingestion_timestamp', current_timestamp())
```

```python
display(bronze_df)
```

### Now we will store the Bronze Layer

```python
bronze_df.write\
    .format("Delta")\
        .mode("overwrite")\
            .save("/Volumes/workspace/default/my_volume/bronze")
```
